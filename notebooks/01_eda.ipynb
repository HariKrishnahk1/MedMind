# 01. Exploratory Data Analysis (EDA)
## Clinical Deterioration Prediction Engine Prototype

> **Research Prototype Disclaimer**: All data analyzed in this notebook is synthetically generated for clinical decision-support research prototyping. No real clinical claims or diagnoses are made.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plot aesthetics
sns.set_theme(style="whitegrid", palette="muted")
%matplotlib inline

# Load synthetic dataset
df = pd.read_csv('../data/raw/synthetic_clinical_data.csv')
print(f"Dataset Loaded: {len(df)} rows, {df['patient_id'].nunique()} unique patients.")
df.head()
```

### 1. Patient Demographics & Observation Distribution

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

sns.histplot(df.groupby('patient_id')['age'].first(), bins=20, ax=axes[0], kde=True, color='teal')
axes[0].set_title('Patient Age Distribution')
axes[0].set_xlabel('Age')

obs_counts = df.groupby('patient_id').size()
sns.histplot(obs_counts, bins=25, ax=axes[1], kde=True, color='coral')
axes[1].set_title('Observations per Patient')
axes[1].set_xlabel('Observation Count')

sns.countplot(x='diagnosis_category', data=df.drop_duplicates(subset=['patient_id']), ax=axes[2], palette='Set2')
axes[2].set_title('Primary Diagnosis Categories')
axes[2].set_xlabel('Diagnosis')

plt.tight_layout()
plt.show()
```

### 2. Missing Value Analysis

```python
missing_pct = (df.isnull().sum() / len(df)) * 100
plt.figure(figsize=(10, 4))
sns.barplot(x=missing_pct.index, y=missing_pct.values, palette='Blues_r')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Missing Percentage (%)')
plt.title('Feature Missingness Rates in Synthetic Clinical Monitoring')
plt.tight_layout()
plt.show()
```
