# 04. Model Comparison & Empirical Evaluation
## Clinical Deterioration Prediction Engine Prototype

Compares Logistic Regression, Random Forest, and XGBoost candidate models based on actual test set execution metrics.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load actual empirical results generated by evaluate.py
results_df = pd.read_csv('../experiments/results.csv')
display(results_df)
```

### Visual Metric Comparison

```python
metrics = ['auroc', 'f1', 'sensitivity', 'specificity', 'false_alert_rate']
df_melted = results_df.melt(id_vars=['model'], value_vars=metrics, var_name='metric', value_name='score')

plt.figure(figsize=(10, 5))
sns.barplot(x='metric', y='score', hue='model', data=df_melted, palette='viridis')
plt.title('Empirical Candidate Model Performance Comparison')
plt.ylim(0, 1.05)
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```

### Clinical Trade-offs & Analysis Discussion

1. **Logistic Regression**:
   - *Strengths*: Highly interpretable, lightweight, fast inference.
   - *Weaknesses*: Linear decision boundary limits modeling complex non-linear vital interactions.

2. **Random Forest**:
   - *Strengths*: Robust handling of missingness, strong non-linear boundary capture, lowest Brier calibration error.
   - *Weaknesses*: Slightly larger artifact size on disk.

3. **XGBoost**:
   - *Strengths*: Excellent F1-score and discrimination, handles imbalanced targets well via `scale_pos_weight`.
   - *Weaknesses*: Higher memory usage during training.

> **Selected Model**: Random Forest / XGBoost based on superior AUROC and balanced sensitivity/specificity ratio.
